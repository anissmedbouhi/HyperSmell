# Entropy visualization

```python
# Load the GSLF data and one representative embedding

epoch = 1000
dataset_name = "gslf"
data_type = "labels"
representation_name = "pom"
batch_size = 195
learning_rate = 0.1
seed = 1 
model_name = "mds"
latent_distance = "poincare"
input_distance = "euclidean"
training_temperature = 0.1
n_neighbors = 20
epsilon = 10.0
optimizer = "PoincareRiemannianAdamOptim"
roi = None
subject = None
pca_dimension = None
pruning_threshold = None

embedding_path = (
    "results_ChemicalSenses_experiments_GSLF/original/"
    f"embeddings_{epoch}_{dataset_name}_{data_type}_"
    f"{representation_name}_{batch_size}_{learning_rate}_{seed}_"
    f"{model_name}_{latent_distance}_{input_distance}_"
    f"{training_temperature}_{n_neighbors}_{epsilon}_{optimizer}_"
    f"{roi}_{subject}_{pca_dimension}_{pruning_threshold}.npy"
)

labels_path = "../../HyperDimRed/data/labels/y_gslf.npy"

X = np.load(embedding_path)
y = np.load(labels_path)

print("Embedding path:", embedding_path)
print("Embedding shape:", X.shape)
print("Descriptor shape:", y.shape)

if X.shape != (y.shape[0], 2):
    raise ValueError(
        f"Embedding shape {X.shape} is incompatible with labels shape {y.shape}."
    )

X = project_inside_poincare_disk(X)

radius = compute_hyperbolic_radius(X)
```

```python
active_label_entropy = compute_active_label_entropy3(
    y,
    normalize=False,
)

active_results = plot_entropy_embedding_and_correlation3(
    X=X,
    entropy=active_label_entropy,
    radius=radius,
    entropy_label="Active label entropy",
    output_prefix=f"gslf_active_label_entropy_seed_{seed}",
    output_directory="gslf_entropy_figures",
)
```

```python
orthogonalized_entropy, pca_scores, pca = (
    compute_orthogonalized_descriptors_entropy3(
        binary_labels=y,
        normalize=False,
        temperature=1.0,
        n_components=y.shape[1],
    )
)

print(
    "Number of PCA components:",
    pca.n_components_,
)

orthogonalized_results = plot_entropy_embedding_and_correlation3(
    X=X,
    entropy=orthogonalized_entropy,
    radius=radius,
    entropy_label="Orthogonalized descriptors entropy",
    output_prefix=f"gslf_orthogonalized_descriptors_entropy_seed_{seed}",
    output_directory="gslf_entropy_figures",
)
```
